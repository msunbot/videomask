```python
video_path = "examples/test_video.mp4"  # update path if needed
```

```python
from videomask.pipeline.segmenter import VideoSegmenter

seg = VideoSegmenter(
    backend="sam3",
    fps=1,
    resize=512,
    max_frames=20,
    backend_kwargs={
        "device": "cuda",
        "text_prompt": "person",
        "score_threshold": 0.0,
    },
)

seg.run(video_path, out_dir="outputs/sam3_demo")
```

```python
import os
from PIL import Image
from IPython.display import display

frames = sorted(os.listdir("outputs/sam3_demo/frames_raw"))
masks = sorted(os.listdir("outputs/sam3_demo/masks"))

print("Frames:", len(frames), "Masks:", len(masks))

display(Image.open("outputs/sam3_demo/frames_raw/" + frames[0]))
display(Image.open("outputs/sam3_demo/masks/" + masks[0]))
```
